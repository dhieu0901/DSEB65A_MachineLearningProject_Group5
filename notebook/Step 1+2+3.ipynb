### **II.2.5. Feature Relationships & Correlation Analysis**

As requested by Step 2 ("Try to understand the relationship between different features... their correlation maybe"), we now analyze the relationships between the numerical variables. The goal is to identify which features are strong predictors for our target (`temp`) and to detect multicollinearity (features that are highly correlated with each other).

A Pearson correlation matrix is computed and visualized as a heatmap.

```python
import numpy as np

numerical_df = df_daily.select_dtypes(include=[np.number])

correlation_matrix = numerical_df.corr()

plt.figure(figsize=(20, 16))
sns.heatmap(
    correlation_matrix, 
    annot=True, 
    fmt='.1f',
    cmap='coolwarm', 
    linewidths=0.5, 
    annot_kws={"size": 8}
)
plt.title('Correlation Matrix of Numerical Weather Features', fontsize=20)
plt.xticks(rotation=45, ha='right')
plt.yticks(rotation=0)
plt.tight_layout()
plt.savefig('hanoi_correlation_heatmap.png')
print("Generated plot: hanoi_correlation_heatmap.png")

print("\n" + "="*50 + "\n")
print("Absolute Correlation with Target Variable (temp):")

corr_with_temp = correlation_matrix['temp'].abs().sort_values(ascending=False)
print(corr_with_temp)
```

**Key Insights from Correlation Analysis:**

1.  **Strongest Predictors for `temp`:**
    * Unsurprisingly, the features most strongly correlated with `temp` are other temperature-based measures: `tempmax` (0.98), `feelslike` (0.97), `tempmin` (0.97), `feelslikemax` (0.96), `feelslikemin` (0.94), and `dew` (0.90). This confirms their relevance but also highlights extreme multicollinearity.
    * A key non-temperature predictor is **`sealevelpressure`**, which has a very strong *negative* correlation (-0.87, as seen in the heatmap). This means as pressure rises, temperature tends to fall (typical of cool, dry air masses in winter).
    * **`solarradiation`** (0.62) and **`solarenergy`** (0.62) show a strong positive correlation, which is intuitive (more sun = higher temperature).
    * `severerisk` (0.65) also shows a high correlation, but this feature is ~73% missing and likely unreliable.

2.  **Multicollinearity (Redundancy):**
    * **Temperature Group:** All `temp` and `feelslike` variables are correlated at > 0.94. Using all of them in a single linear model would be highly problematic.
    * **Solar Group:** `solarradiation` and `solarenergy` are almost perfectly correlated (1.0). One of these must be dropped.
    * **Precipitation Group:** `precip`, `precipprob`, and `precipcover` are moderately correlated, suggesting they capture related aspects of precipitation.
    * `snow` and `snowdepth` show `NaN` (Not a Number) in the correlation with `temp`. This is because they have zero variance (the value is always 0 for Hanoi), making them useless for prediction. These columns should be dropped.

3.  **Weak Predictors:**
    * Features like `moonphase` (0.009), `precip` (0.09), `windspeed` (0.15), and `humidity` (0.13) show a very weak *linear* relationship with the daily average temperature. This doesn't mean they are useless (they might have non-linear relationships), but they are not strong direct drivers.

### **II.2.6. How Features Combine to Predict Temperature (Initial Hypotheses)**

Based on the target variable analysis and correlation matrix, we can formulate several initial hypotheses about what drives Hanoi's temperature. These hypotheses will guide our feature engineering and modeling strategies.

**Hypothesis 1: Seasonality is the Primary Driver.**
* **Evidence:** The time-series plot shows a powerful, dominant annual cycle, and the bimodal distribution confirms two distinct "seasons."
* **Implication:** The most important predictive features will not be the raw columns, but rather **engineered time-based features** that explicitly model this seasonality (e.g., `day_of_year`, `month`, sine/cosine transformations of the date).

**Hypothesis 2: Temperature is Strongly Autoregressive ("Sticky").**
* **Evidence:** Weather is a continuous physical process. The temperature on any given day is highly dependent on the temperature of the previous day(s).
* **Implication:** **Lag features** (e.g., `temp` from 1 day ago, 7 days ago) will be extremely powerful predictors. The task can be framed as "Given the temperature yesterday and the seasonal context, what is the temperature today?"

**Hypothesis 3: Atmospheric & Solar Conditions are Key Exogenous Drivers.**
* **Evidence:** After accounting for seasonal cycles, the remaining day-to-day variance will be explained by other physical factors. The correlation matrix identified two strong, non-temperature-based drivers:
    1.  **`sealevelpressure`** (strong negative correlation): Higher pressure likely indicates cool, dry air, leading to lower temperatures.
    2.  **`solarradiation` / `solarenergy`** (strong positive correlation): More solar energy directly increases temperature.
* **Implication:** These features should be included as primary exogenous variables in our model.

**Hypothesis 4: Text Features Provide Actionable Context.**
* **Evidence:** The project brief specifically mentions leveraging text features. Columns like `conditions` and `icon` (e.g., "Rain, Overcast", "Clear") are categorical summaries of complex events.
* **Implication:** A day labeled "Overcast" will likely have a lower `tempmax` and `temp` than a "Clear" day, even if other numerical features are similar. These categorical features must be encoded (e.t., One-Hot Encoding) as they capture information that `cloudcover` (a percentage) might not fully represent.

**Hypothesis 5: Some Features are Redundant or Noise.**
* **Evidence:**
    1.  **Redundant:** `tempmax`, `tempmin`, `feelslike`, etc., are *collinear* with `temp`. `solarenergy` is a direct copy of `solarradiation`.
    2.  **Zero-Variance:** `snow` and `snowdepth` are always 0.
    3.  **Missing:** `severerisk` is ~73% missing.
    4.  **Irrelevant:** `moonphase` has a 0.009 correlation and is physically unlikely to drive daily temperature.
* **Implication:** These features will be cleaned or removed in the Data Processing (Step 3) to reduce model complexity and prevent errors.

### **III.2. Feature Type Identification**

This step executes the processing plan defined above. We load the raw data, drop the 9 identified columns, and ensure the `datetime` column is correctly formatted.

```python
# Constant and Zero-Variance Columns
def find_constant_columns(df):
    """Find columns with only one unique (non-null) value."""
    # This also finds numeric columns with zero variance
    return [col for col in df.columns if df[col].nunique(dropna=True) <= 1]

url = "https://raw.githubusercontent.com/dhieu0901/DSEB65A_MachineLearningProject_Group5/main/data/Hanoi%20Daily%2010%20years.csv"
hanoi_daily_df = pd.read_csv(url)
# find all constant columns
constant_cols = find_constant_columns(hanoi_daily_df)
print("\nConstant or Zero-Variance Columns Detected:")
print(f"Dataset Shape: {hanoi_daily_df.shape}")
print(constant_cols)

# (domain knowledge) other potential cols that should be dropped for future refactor
other_cols_to_drop = ['address', 'resolvedAddress', 'source', 'latitude', 'longitude', 'severerisk', 'stations']
print(hanoi_daily_df.head())

# combine the lists and remove duplicates -> use set
all_cols_to_drop = list(set(constant_cols + other_cols_to_drop))

hanoi_daily_df.info()
print("\nFull list of columns to be dropped:")
print(all_cols_to_drop)

# 4. drop the columns that actually exist in the dataframe
# PLEASE note that from now on we only process with cleaned_df for better later debugging
cleaned_df = hanoi_daily_df.drop(columns=[c for c in all_cols_to_drop if c in hanoi_daily_df.columns], errors='ignore')
df_processed = cleaned_df

# ensure datetime columns are properly converted
if 'datetime' in df_processed.columns:
    df_processed['datetime'] = pd.to_datetime(df_processed['datetime'])
if 'sunrise' in df_processed.columns:
    df_processed['sunrise'] = pd.to_datetime(df_processed['sunrise'])
if 'sunset' in df_processed.columns:
    df_processed['sunset'] = pd.to_datetime(df_processed['sunset'])

if 'conditions' in df_processed.columns:
    conditions_dummies = df_processed['conditions'].str.get_dummies(sep=', ')
    df_processed = pd.concat([df_processed, conditions_dummies], axis=1)
    df_processed.drop('conditions', axis=1, inplace=True)
    if 'description' in df_processed.columns:
        df_processed.drop('description', axis=1, inplace=True)

if 'icon' in df_processed.columns:
    icon_dummies = pd.get_dummies(df_processed['icon'], prefix='icon', dummy_na=False)
    df_processed = pd.concat([df_processed, icon_dummies], axis=1)
    df_processed.drop('icon', axis=1, inplace=True)

if 'preciptype' in df_processed.columns:
    df_processed['preciptype'] = df_processed['preciptype'].fillna('None')
    precip_dummies = pd.get_dummies(df_processed['preciptype'], prefix='preciptype', dummy_na=False)
    df_processed = pd.concat([df_processed, precip_dummies], axis=1)
    df_processed.drop('preciptype', axis=1, inplace=True)

print("After encoding: shape =", df_processed.shape)
print("Sample columns (tail):", list(df_processed.columns)[-15:])
df_processed.info()
```

### **III.6. Data Normalization and Transformation**

**Discussion on Methodology**

Data normalization (or scaling) is a crucial preprocessing step for many machine learning algorithms, especially those that are sensitive to the magnitude of features (e.g., Linear Regression, SVMs, Neural Networks). The goal is to transform numerical features to a common scale, typically:
* **Normalization (Min-Max Scaling):** Rescales features to a fixed range, usually 0 to 1.
* **Standardization (Z-score Scaling):** Rescales features to have a mean of 0 and a standard deviation of 1.

However, it is a critical methodological error to apply scaling to the *entire dataset* at this stage.

**Risk of Data Leakage:**
If we were to scale the data *before* splitting it into training, validation, and testing sets, the scaler (e.g., `MinMaxScaler`) would learn the `min` and `max` values from the *entire 10-year dataset*. This means information from the future (the test set) would "leak" into the past (the training set), giving the model an unrealistic advantage and leading to an overly optimistic performance evaluation.

**Correct Approach:**
The correct, non-leaking approach is to:
1.  Split the data (e.g., into train/test sets).
2.  **Fit** the scaler (`.fit()`) **only** on the **training data**.
3.  Use that *fitted* scaler to **transform** (`.transform()`) both the training data and the test data.

**Conclusion:**
Therefore, this data processing step **will not** apply a global normalization. This transformation will be handled as a key component within the modeling pipeline (in Step 5), applied *after* the data has been split to ensure the integrity of the model's evaluation.
